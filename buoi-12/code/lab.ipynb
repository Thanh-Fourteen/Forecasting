## Bước 1 — Nhìn phổ trước khi lọc (mục 4.1, 4.2)

```python
%matplotlib inline
import warnings

import khu_nhieu as kn
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

warnings.simplefilter("ignore")
nam_gio = pd.Series([10, 12, 14, 30, 16.0])
print(pd.DataFrame({"điện": nam_gio, "trailing": nam_gio.rolling(3).mean().round(2),
                    "centered": nam_gio.rolling(3, center=True).mean().round(2)}))

df = kn.doc_cam_bien()
fig, truc = plt.subplots(1, 2, figsize=(11, 3.2))
for ax, cot in zip(truc, ["Appliances", "T2"], strict=True):
    f, P = kn.pho(df[cot].interpolate())
    ax.loglog(f[1:], P[1:])
    ax.set_title(cot)
    ax.set_xlabel("chu kỳ/giờ")
    print(cot, "| chu kỳ mạnh nhất (giờ):", round(1 / f[np.argmax(P)], 2),
          "| phần năng lượng ở chu kỳ < 1 giờ:", round(P[f > 1].sum() / P[1:].sum(), 3))
```

## Bước 2 — Aliasing (mục 4.3)

Sửa `ha_mau` rồi chạy lại ô này.

```python
t = np.arange(6000) / kn.FS                                   # giờ, 10 phút một mẫu
for tan_so in (1.4, 1.5):
    y = 20 + 2 * np.sin(2 * np.pi * t / 24) + 0.5 * np.sin(2 * np.pi * tan_so * t)
    tho, sach = kn.ha_mau(y, 6, loc_truoc=False), kn.ha_mau(y, 6, loc_truoc=True)
    f1, P1 = kn.pho(tho, fs=1.0, nperseg=512)
    f2, P2 = kn.pho(sach, fs=1.0, nperseg=512)
    print(f"dao động {tan_so} chu kỳ/giờ → công suất ở chu kỳ 2,5 giờ: thô {kn.cong_suat_quanh(f1, P1, 2.5):.1f}, "
          f"lọc trước {kn.cong_suat_quanh(f2, P2, 2.5):.1f}")
```
