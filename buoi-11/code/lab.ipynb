## Bước 1 — Ngưỡng toàn chuỗi so với cửa sổ trượt (mục 4.2, 4.3)

```python
%matplotlib inline
import warnings

import bat_thuong as bt
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import ruptures as rpt

warnings.simplefilter("ignore")
tong, tet, hk = bt.doc_tong_vi(), bt.doc_bai_tet(), bt.doc_hang_khong()

muoi = pd.Series([10, 12, 11, 13, 12, 50, 11, 12, 13, 12.0])
print("z-score:", ((muoi - muoi.mean()) / muoi.std()).round(2).tolist())
print("điểm MAD:", bt.mad_score(muoi).round(2).tolist())

print(bt.so_sanh_bat(tong).to_string(index=False))
print(bt.masking(tong))
```

## Bước 2 — Tết: sự kiện thật (mục 4.4)

Sửa `xu_ly_ngoai_lai` rồi chạy lại ô này.

```python
print(bt.mat_bao_nhieu_dinh_tet(tet).to_string(index=False))
dinh = bt.dinh_tet(tet)
xl = bt.xu_ly_ngoai_lai(tet, bo_qua_su_kien=dinh)
print({"số mốc trước": len(tet), "số mốc sau": int(xl["sach"].notna().sum()),
       "đã sửa": int(xl["da_sua"].sum()), "bị xoá": int(xl["bi_xoa"].sum()),
       "đỉnh Tết giữ nguyên": bool(np.allclose(xl.loc[dinh, "sach"], tet.loc[dinh]))})
```

## Bước 3 — PELT và penalty (mục 4.5)

Sửa `diem_gay` (lấy log khi `log=True`) rồi chạy lại ô này.

```python
muoi_gay = np.array([10, 11, 10, 9, 10, 20, 21, 19, 20, 20.0])
print("10 số, penalty 3 ln 10:", rpt.Pelt(model="l2", min_size=2).fit(muoi_gay).predict(pen=3 * np.log(10)))

print("mức gốc:", len(bt.diem_gay(hk, log=False)), "điểm gãy | log:", len(bt.diem_gay(hk, log=True)), "điểm gãy")
quet = bt.quet_penalty(hk)
fig, ax = plt.subplots(figsize=(6, 2.8))
ax.plot(quet["pen/log n"], quet["số điểm gãy"], "o-")
ax.set_xlabel("penalty / ln n")
ax.set_ylabel("số điểm gãy")
plt.show()
moc = bt.diem_gay_on_dinh(hk)
print(bt.do_lon_gay(hk, moc).to_string(index=False))
```
