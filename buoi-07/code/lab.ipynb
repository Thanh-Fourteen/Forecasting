## Bước 5 — Chuỗi thật

Với mỗi chuỗi, viết một dòng: ô nào của bảng 2 × 2, số lần sai phân, bằng chứng nào.

```python
def in_kd(ten, y, ct=False):
    kq = {k: round(v, 3) if isinstance(v, float) else v for k, v in tt.kiem_dinh(y, co_xu_huong=ct).items()}
    print(f"{ten:32s}", "ct" if ct else "c ", kq, "→", tt.ket_luan(y, ct))


log_gdp = np.log(tt.doc_gdp())
tang = log_gdp.diff().dropna() * 100
log_ip = np.log(tt.doc_san_luong())
in_kd("log GDP", log_gdp, ct=True)
in_kd("tăng trưởng GDP, cả giai đoạn", tang)
in_kd("tăng trưởng GDP, 1985–2019", tang["1985":"2019"])
in_kd("log sản lượng công nghiệp", log_ip)
in_kd("log sản lượng công nghiệp", log_ip, ct=True)
in_kd("sai phân log sản lượng", log_ip.diff().dropna())
print("số lần sai phân log GDP:", tt.so_lan_sai_phan(log_gdp))
print("sai phân lần hai GDP:", {k: round(v, 3) for k, v in tt.dau_hieu_sai_phan_thua(tang).items()})
```
